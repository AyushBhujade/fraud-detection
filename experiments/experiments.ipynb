```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv(r"..\data\creditcard.csv")
```

```python
df.head()
```

```python
df["Class"].value_counts()
```

```python
x=df.drop(["Class"],axis=1)
y=df["Class"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
y_test.value_counts()
```

```python
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
from imblearn.over_sampling import SMOTE
```

```python
smote = SMOTE()
X_train_res, y_train_res = smote.fit_resample(X_train_scaled, y_train)
```

```python
y_train_res.value_counts()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import StackingClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.svm import SVC
from sklearn.metrics import recall_score,f1_score,confusion_matrix,precision_score
```

```python

import dagshub
import mlflow

mlflow.set_tracking_uri("https://dagshub.com/ayushbhujade2005/fraud-detection.mlflow")
dagshub.init(repo_owner='ayushbhujade2005', repo_name='fraud-detection', mlflow=True)
#mlflow.set_experiment("SMOTE technique")
```

```python
import os,time
from sklearn.metrics import precision_score
def train_model(classifier,X_train,X_test,y_train,y_test):
    with mlflow.start_run():
        start_time=time.time()
        mlflow.log_param("imbalance technique","SMOTE")
        mlflow.log_param("scaler","StandardScaler")
        model=classifier()
        print(f"training the {classifier} model")
        model.fit(X_train,y_train)
        print(f"training completed now model making predictions")
        mlflow.log_param("model",classifier)
        y_pred=model.predict(X_test)
        
        print("calculation evaluation matrix")
        recall=recall_score(y_pred=y_pred,y_true=y_test)
        precision=precision_score(y_pred=y_pred,y_true=y_test)
        matrix=confusion_matrix(y_pred=y_pred,y_true=y_test)
        f1=f1_score(y_pred=y_pred,y_true=y_test)
        
        mlflow.log_metric("recall_score",recall)
        mlflow.log_metric("precision_score",precision)
        mlflow.log_metric("f1_score",f1)
        
        print("logging and saving the model")
        
        mlflow.sklearn.log_model(classifier,"model")
        
        end_time=time.time()
        
        print(f"model training and model evaluation completed in {end_time-start_time:.2f} seconds")
        print(f"f1_score---------->{f1}")
        print(f"precision---------->{precision}")
        print(f"confusion_matrix---------->{matrix}")
        print(f"recall---------->{recall}")
```

```python
train_model(X_train=X_train_res,X_test=X_test,y_train=y_train_res,y_test=y_test,classifier=DecisionTreeClassifier)
```
